  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>

```python
#202432061_Muhammad Yusuf Khairan
import re
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432061_MuhammadYusufKhairan
# 1. Muat dataset ulasan Lazada (20191002-reviews.csv dari Kaggle: Lazada Indonesian Reviews)
DATA_PATH = '20191002-reviews.csv'
df = pd.read_csv(DATA_PATH)
print('Jumlah baris awal                 :', len(df))

# 2. Pembersihan data
df = df.drop_duplicates()
df = df.dropna(subset=['reviewContent'])
df = df[df['rating'] != 3].copy()
print('Setelah dedup, tanpa kosong & netral:', len(df))

# 3. Pelabelan sentimen (klasifikasi biner)
df['label'] = (df['rating'] >= 4).astype(int)

# 4. Case folding, pembersihan karakter, dan normalisasi teks
def clean_text(text):
    text = str(text).lower()
    text = re.sub(r'http\S+|www\.\S+', ' ', text)
    text = re.sub(r'[^a-z\s]', ' ', text)
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    return re.sub(r'\s+', ' ', text).strip()

df['clean_text'] = df['reviewContent'].apply(clean_text)
df = df[df['clean_text'].str.len() > 0].copy()
print('Negatif:', (df['label'] == 0).sum(), '| Positif:', (df['label'] == 1).sum())

# 5. Penyeimbangan kelas (undersampling) agar kelas mayoritas tidak mendominasi
n_per_class = df['label'].value_counts().min()
pos = df[df['label'] == 1].sample(n=n_per_class, random_state=42)
neg = df[df['label'] == 0].sample(n=n_per_class, random_state=42)
df_bal = pd.concat([pos, neg]).sample(frac=1, random_state=42).reset_index(drop=True)
print('Jumlah data seimbang (total):', len(df_bal))

# 6. Pembagian data latih dan data uji (80:20, stratified agar proporsi kelas tetap seimbang)
X_text = df_bal['clean_text'].tolist()
y_labels = df_bal['label'].to_numpy()
X_train_txt, X_test_txt, y_train, y_test = train_test_split(X_text, y_labels, test_size=0.2, random_state=42, stratify=y_labels)   # bagi data
print('Data latih:', len(X_train_txt), '| Data uji:', len(X_test_txt))

# 7. Tokenisasi dan vocabulary (dibangun HANYA dari data latih agar tidak terjadi kebocoran data)
VOCAB_SIZE = 10000
MAX_LEN = 50
counter = Counter(w for text in X_train_txt for w in text.split())
itos = ['<PAD>', '<UNK>'] + [w for w, _ in counter.most_common(VOCAB_SIZE - 2)]
stoi = {w: i for i, w in enumerate(itos)}
print('Ukuran vocabulary:', len(itos))

def encode(text):
    ids = [stoi.get(w, 1) for w in text.split()]
    ids = ids[:MAX_LEN]
    return [0] * (MAX_LEN - len(ids)) + ids

# 8. Konversi ke array NumPy lalu ke PyTorch Tensor
X_train = np.array([encode(t) for t in X_train_txt], dtype=np.int64)
X_test = np.array([encode(t) for t in X_test_txt], dtype=np.int64)
X_train_t = torch.tensor(X_train, dtype=torch.long)
y_train_t = torch.tensor(y_train, dtype=torch.float32)
X_test_t = torch.tensor(X_test, dtype=torch.long)
y_test_t = torch.tensor(y_test, dtype=torch.float32)
print('Shape X_train:', tuple(X_train_t.shape), '| Shape X_test:', tuple(X_test_t.shape))
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432061_MuhammadYusufKhairan
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embed_dim=128, hidden_size=128, dropout=0.3):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.lstm = nn.LSTM(embed_dim, hidden_size, batch_first=True)
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_size, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        emb = self.embedding(x)
        _, (h_n, _) = self.lstm(emb)
        out = self.dropout(h_n[-1])
        out = self.fc(out)
        out = self.sigmoid(out)
        return out.squeeze(1)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432061_MuhammadYusufKhairan
# 1. Inisiasi hyperparameter
torch.manual_seed(42)
np.random.seed(42)
BATCH_SIZE = 64
epochs = 15
learning_rate = 0.001
embed_dim = 128
hidden_size = 128
dropout = 0.3

# 2. Inisiasi model, fungsi loss, dan optimizer
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = SequenceModel(len(itos), embed_dim, hidden_size, dropout).to(device)
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. DataLoader: membagi data latih menjadi batch dan mengacaknya setiap epoch
train_loader = DataLoader(TensorDataset(X_train_t, y_train_t), batch_size=BATCH_SIZE, shuffle=True)

# 4. Simpan riwayat loss dan akurasi pelatihan
train_losses = []
train_accs = []

# 5. Training loop
for epoch in range(epochs):
    model.train()
    total_loss, correct, total = 0.0, 0, 0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        preds = model(xb)
        loss = criterion(preds, yb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        total_loss += loss.item() * xb.size(0)
        correct += ((preds >= 0.5).float() == yb).sum().item()
        total += xb.size(0)
    train_losses.append(total_loss / total)
    train_accs.append(correct / total)
    print(f'Epoch {epoch + 1:02d}/{epochs} | Loss: {train_losses[-1]:.4f} | Akurasi latih: {train_accs[-1]:.4f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432061_MuhammadYusufKhairan
# 1. Evaluasi model pada data uji
model.eval()
with torch.no_grad():
    test_probs = model(X_test_t.to(device)).cpu().numpy()
test_preds = (test_probs >= 0.5).astype(int)
print(classification_report(y_test, test_preds, target_names=['Negatif', 'Positif'], digits=4))

# 2. Grafik penurunan loss dan kenaikan akurasi selama pelatihan
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
axes[0].set_title('Grafik Penurunan Loss Selama Pelatihan')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].grid(True, linestyle='--', alpha=0.5)
axes[0].legend()
axes[1].plot(train_accs, color='#1FA4B1', linewidth=2, label='Training Accuracy')
axes[1].set_title('Grafik Akurasi Selama Pelatihan')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Akurasi')
axes[1].grid(True, linestyle='--', alpha=0.5)
axes[1].legend()
plt.show()

# 3. Grafik perbandingan data aktual vs hasil prediksi, dan distribusi skor sentimen
labels_name = ['Negatif', 'Positif']
actual_counts = [int((y_test == 0).sum()), int((y_test == 1).sum())]
pred_counts = [int((test_preds == 0).sum()), int((test_preds == 1).sum())]
x = np.arange(len(labels_name))
width = 0.35
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].bar(x - width / 2, actual_counts, width, color='#114F6A', label='Data Aktual')
axes[0].bar(x + width / 2, pred_counts, width, color='#1FA4B1', label='Hasil Prediksi')
axes[0].set_xticks(x)
axes[0].set_xticklabels(labels_name)
axes[0].set_title('Data Aktual vs Hasil Prediksi (Data Uji)')
axes[0].set_ylabel('Jumlah ulasan')
axes[0].set_ylim(0, max(actual_counts + pred_counts) * 1.25)
axes[0].legend()
axes[1].hist(test_probs[y_test == 1], bins=30, color='#114F6A', alpha=0.7, label='Aktual Positif')
axes[1].hist(test_probs[y_test == 0], bins=30, color='#1FA4B1', alpha=0.7, label='Aktual Negatif')
axes[1].axvline(0.5, color='black', linestyle='--', label='Ambang 0.5')
axes[1].set_title('Distribusi Skor Prediksi Sentimen (Probabilitas Positif)')
axes[1].set_xlabel('Skor probabilitas positif')
axes[1].set_ylim(0, axes[1].get_ylim()[1] * 1.25)
axes[1].set_ylabel('Jumlah ulasan')
axes[1].legend()
plt.tight_layout()
plt.show()

# 4. Contoh hasil prediksi pada 5 ulasan acak dari data uji
idx = np.random.RandomState(0).choice(len(X_test_txt), size=5, replace=False)
for i in idx:
    aktual = 'Positif' if y_test[i] == 1 else 'Negatif'
    prediksi = 'Positif' if test_preds[i] == 1 else 'Negatif'
    print(f'[aktual: {aktual} | prediksi: {prediksi} | skor: {test_probs[i]:.3f}] {X_test_txt[i][:90]}')
```
